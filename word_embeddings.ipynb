```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
```

```python
wrd_to_idx = {"i" : 0, "love" : 1, "eating" : 2, "and" : 3, "sleeping" : 4}
embeddings = nn.Embedding(5,7) #Embedding(<no. of words you have>, <each word will be represented by this much values in a vector>)

word_index = torch.tensor([wrd_to_idx["love"]])
print(word_index)
```

```python
love = embeddings(word_index)
print(love)
print(love.shape)
```

```python
all_indices = torch.tensor([w for w in range(5)], dtype = torch.long)
all_words = embeddings(all_indices)
print(all_words)
print(all_words.shape)
```
